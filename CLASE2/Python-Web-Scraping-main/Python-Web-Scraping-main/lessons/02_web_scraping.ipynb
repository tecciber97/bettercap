# Extracción y análisis de HTML


Para extraer y analizar HTML correctamente, seguiremos los siguientes 4 pasos:
1. Realizar una solicitud GET
2. Analizar la página con Beautiful Soup
3. Buscar elementos HTML
4. Obtener los atributos y el texto de estos elementos

## Paso 1: Realizar una solicitud GET para obtener el HTML de una página

Podemos utilizar la biblioteca de Solicitudes para:

1. Realizar una solicitud GET a la página y
2. Leer el código HTML de la página web.

El proceso de realizar una solicitud y obtener un resultado es similar al del flujo de trabajo de la API web. Sin embargo, ahora realizamos una solicitud directamente al sitio web y tendremos que analizar el HTML nosotros mismos. Esto contrasta con recibir datos organizados en una salida más sencilla en formato `JSON` o `XML`.

```python
# Make a GET request
req = requests.get('http://www.ilga.gov/senate/default.asp')
# Read the content of the server’s response
src = req.text
# View some output
print(src[:1000])
```

## Paso 2: Analiza la página con Beautiful Soup

Ahora, usamos la función  `BeautifulSoup` para analizar la respuesta y convertirla en un árbol HTML. Esto devuelve un objeto (llamado **objeto sopa**) que contiene todo el HTML del documento original.

Si encuentra un error sobre una biblioteca de analizador, asegúrese de haber instalado el paquete `lxml` para proporcionar a Beautiful Soup las herramientas de análisis necesarias.

```python
# Parse the response into an HTML tree
soup = BeautifulSoup(src, 'lxml')
# Take a look
print(soup.prettify()[:1000])
```

La salida se ve bastante similar a la anterior, pero ahora está organizada en un objeto `soup` que nos permite recorrer la página más fácilmente.

## Paso 3: Buscar elementos HTML

Beautiful Soup cuenta con varias funciones para encontrar componentes útiles en una página. Beautiful Soup te permite encontrar elementos por:

1. Etiquetas HTML
2. Atributos HTML
3. Selectores CSS
Busquemos primero las **etiquetas HTML**.
 

La función `find_all` busca en el árbol `soup` todos los elementos con una etiqueta HTML específica y los devuelve.

¿Qué hace el siguiente ejemplo?

```python
# Find all elements with a certain tag
a_tags = soup.find_all("a")
print(a_tags[:10])
```

Dado que `find_all()` es el método más popular en la API de búsqueda de Beautiful Soup, puedes usar un atajo. Si tratas el objeto BeautifulSoup como si fuera una función, es lo mismo que llamar a `find_all()` en ese objeto.

Estas dos líneas de código son equivalentes:

```python
a_tags = soup.find_all("a")
a_tags_alt = soup("a")
print(a_tags[0])
print(a_tags_alt[0])
```

¿Cuantos enlaces obtuvimos?

```python
print(len(a_tags))
```

¡Eso es muchísimo! Muchos elementos de una página tendrán la misma etiqueta HTML. Por ejemplo, si buscas todo con la etiqueta `a`, probablemente obtendrás más resultados, muchos de los cuales quizás no quieras. Recuerda que la etiqueta `a` define un hipervínculo, por lo que normalmente encontrarás muchos en cualquier página.

¿Qué sucedería si quisiéramos buscar etiquetas HTML con ciertos atributos, como clases CSS específicas?

Podemos hacerlo añadiendo un argumento adicional a `find_all`. En el siguiente ejemplo, buscamos todas las etiquetas `a` y luego las filtramos con `class_="sidemenu"`.

```python
# Get only the 'a' tags in 'sidemenu' class
side_menus = soup("a", class_="sidemenu")
side_menus[:5]
```

Una forma más eficiente de buscar elementos en un sitio web es mediante un selector CSS. Para ello, debemos usar un método diferente llamado `select()`. Simplemente pase una cadena a `.select()` para obtener todos los elementos con esa cadena como un selector CSS válido.

## 🥊 Desafío práctico: Crear una función de scraping

Queremos scraping las páginas web correspondientes a los proyectos de ley patrocinados por cada proyecto de ley.

Escribir una función llamada `get_bills(url)` para analizar la URL de un proyecto de ley. Esto implica:

  - solicitando la URL mediante el <a href="http://docs.python-requests.org/en/latest/">`requests`</a> Biblioteca - Utilizando las funciones de la `BeautifulSoup` Biblioteca para encontrar todos los `<td>` e
elementos con la clase `billlist`
  - - devuelve un _list_ of tuples, cada una con:
      - Descripción (2.ª columna)
      - Cámara (S o H) (3.ª columna)
      - Última acción (4.ª columna)
      - Fecha de la última acción (5.ª columna)
      
Esta función se ha completado parcialmente. Complete el resto.

```python
import requests
from bs4 import BeautifulSoup
```

```python
def get_bills(url):
    # Realizamos la solicitud HTTP
    response = requests.get(url)
    
    # Si la respuesta es exitosa (status code 200)
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')

        # Encuentra todos los elementos <td> con la clase 'billlist'
        bill_entries = soup.select('td.billlist')

        # Lista para almacenar los resultados
        bills = []

        # Recorremos todas las filas de los proyectos de ley
        for entry in bill_entries:
            # Obtener todas las celdas <td> en la fila
            columns = entry.find_all('td')

            # Asegúrate de que hay suficientes columnas en la fila (debe haber 5)
            if len(columns) >= 5:
                # Extraemos los valores de las columnas necesarias
                description = columns[1].get_text(strip=True)
                chamber = columns[2].get_text(strip=True)
                last_action = columns[3].get_text(strip=True)
                action_date = columns[4].get_text(strip=True)

                # Agregar la tupla con los datos extraídos a la lista de resultados
                bills.append((description, chamber, last_action, action_date))
        
        # Devolver la lista de tuplas con los proyectos de ley
        return bills
    else:
        print(f"Error al obtener la página: {response.status_code}")
        return []
```

### Extraer todos los proyectos de ley

Finalmente, cree un diccionario `bills_dict` que asigne un número de distrito (la clave) a una lista de proyectos de ley (el valor) provenientes de ese distrito. Puede hacerlo recorriendo en bucle todos los miembros del senado en `members_dict` y llamando a `get_bills()` para cada una de las URL de sus proyectos de ley asociados.

**NOTA:** Por favor, llame a la función `time.sleep(1)` en cada iteración del bucle para no destruir el sitio web del estado.

```python
import time
import requests
from bs4 import BeautifulSoup
```

```python
def get_bills(url):
    # Realizamos la solicitud HTTP
    response = requests.get(url)
    
    # Si la respuesta es exitosa (status code 200)
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')

        # Aquí necesitas encontrar cómo se estructuran los proyectos de ley en el HTML
        # Por ejemplo, podríamos estar buscando todos los enlaces de proyectos de ley en una lista
        bills = soup.select('a.bill-link')  # Ajusta el selector según la estructura del HTML

        # Extraemos los títulos o números de los proyectos de ley
        bill_titles = [bill.get_text() for bill in bills]

        return bill_titles
    else:
        print(f"Error al obtener la página: {response.status_code}")
        return []
```

```python
# Diccionario para almacenar los proyectos de ley por número de distrito
bills_dict = {}

# Supongamos que tienes un diccionario 'members_dict' con el número de distrito como clave
# y la URL de los proyectos de ley del miembro del senado como valor
members_dict = {
    1: 'http://www.ilga.gov/senate/member/1/bills',
    2: 'http://www.ilga.gov/senate/member/2/bills',
    3: 'http://www.ilga.gov/senate/member/3/bills',
    # Agrega más miembros con sus respectivas URLs
}

# Recorremos todos los miembros
for district, url in members_dict.items():
    print(f"Extrayendo proyectos de ley para el distrito {district}...")
    
    # Llamamos a get_bills() para obtener los proyectos de ley de la URL
    bills = get_bills(url)
    
    # Almacenamos los proyectos de ley en el diccionario bills_dict
    bills_dict[district] = bills
    
    # Hacemos una pausa de 1 segundo entre cada iteración para no sobrecargar el servidor
    time.sleep(1)

# Imprimir el resultado
print(bills_dict)
```
